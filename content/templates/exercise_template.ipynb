# 知識点のタイトル — 練習問題

> **【練習問題 テンプレート】**（このセルは削除してよい）
>
> - 1問 = `## 問題 <問題ID>` で始まる markdown セル。問題 ID は全体で一意。
>   **新しい問題は `## 問題` とだけ書けばよい。** `uv run python -m tools.edu_content assign-ids content/` を実行すると ID（例: `loop_03-7f3a9c`）が自動で付き、このファイルに書き戻される。
> - 見出しの直後に属性を箇条書きで書く：
>   - `形式`：`choice`（選択問題）または `code`（コード問題）
>   - `知識点`：複数書ける。**先頭が主知識点**。省略時は `default_knowledge_point`
>   - `難易度`：1〜9。省略時は `default_difficulty` → 主知識点の難易度
>   - `キーワード`：`` `a` `b` `` または `a, b`
>   - `タグ`（任意）：分類用のラベル。例：`バグ取り`、`穴埋め`、`複数選択`。表示・採点には影響しない

> - 選択問題は `### 正解` に正解の記号（`B`、複数正解なら `A, C`）、`### 解説` に解説を書く。
> - コード問題は `### 解説`（任意）を書き、続くセルに `# @starter`（初期コード）、`# @solution`（模範解答）、`@tests`（テストケース）を置く。
>   - バグ取り：`@starter` にバグのあるプログラムを置き、問題文にエラー出力などを書く
>   - 穴埋め：`@starter` にプログラムの一部を書いておき、残りを学生に書かせる

## 問題 tmpl_01_q01

- 形式: choice
- 難易度: 1
- キーワード: `キーワード1` `キーワード2`
- タグ: 基本

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
x = 10
print(x)
```

- A. `10`
- B. `x`
- C. `'10'`
- D. エラーになる

### ヒント

1. `print()` に変数を渡すと、変数の中身が表示される。

### 正解

A

### 解説

## 問題 tmpl_01_c01

- 形式: code
- 難易度: 2
- キーワード: `input` `print`

### 問題文

整数 `n` を入力として受け取り、`n` の2倍を出力するプログラムを書け。

### ヒント

1. `input()` で受け取った値は文字列なので `int()` で整数に変換する。

### 解説

標準入力 → 標準出力で採点する問題（`mode: stdio`）の例。

````python
# @starter
n = int(input())
# ここにコードを書く
````

````python
# @solution
n = int(input())
print(n * 2)
````

## 問題 tmpl_01_c02

- 形式: code
- 難易度: 1
- タグ: バグ取り

### 問題文

キーボードから入力された値の3乗を出力するプログラムを作成したが、実行すると次のエラーになる。プログラムを訂正せよ。

```
TypeError: unsupported operand type(s) for ^: 'float' and 'int'
```

### 解説

バグ取り問題の例。`@starter` にバグのあるプログラムを置く。Python のべき乗は `**`（`^` はビット演算の XOR）。

````python
# @starter
x = float(input())
print(x ^ 3)
````

````python
# @solution
x = float(input())
print(x ** 3)
````

```yaml
# @tests
mode: stdio
cases:
  - input: "3"
    expected: "27.0"
    public: true
  - input: "-2"
    expected: "-8.0"
```

## 問題 tmpl_02_c01

- 形式: code
- 知識点: tmpl_02, tmpl_01
- 難易度: 2
- タグ: 穴埋め

### 問題文

関数 `double(x)` を完成させよ。`x` の2倍を返す。

### 解説

関数を呼び出して戻り値で採点する問題（`mode: function`）の例。初期コードに関数の枠だけを書いておく「穴埋め」の形にしている。知識点を複数書くと、先頭（`tmpl_02`）が主知識点になる。

````python
# @starter
def double(x):
    # ここにコードを書く
    pass
````

````python
# @solution
def double(x):
    return x * 2
````

```yaml
# @tests
# mode: function … コードを実行したあと call を評価し、戻り値を expected と == で比較する
mode: function
cases:
  - call: double(3)
    expected: 6
    public: true
  - call: double(0)
    expected: 0
  - call: double(-2.5)
    expected: -5.0
```

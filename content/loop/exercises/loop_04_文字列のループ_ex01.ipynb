# 文字列のループ — 練習問題

## 問題 loop_04_q01

- 形式: choice
- 難易度: 1
- キーワード: `for ch in 文字列`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for ch in 'abc':
    print(ch)
```

- A. `abc`
- B. `a` / `b` / `c`
- C. `c` / `b` / `a`
- D. エラーになる

### 正解

B

### 解説

文字列を for で回すと、1文字ずつ順に取り出される。

## 問題 loop_04_q02

- 形式: choice
- 難易度: 2
- キーワード: `for ch in 文字列` `文字列の連結`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
s = ''
for ch in 'cat':
    s = ch + s
print(s)
```

- A. `cat`
- B. `tac`
- C. `c` / `a` / `t`
- D. `ccaatt`

### 正解

B

### 解説

毎回、取り出した文字を `s` の**前**に付けるので、順番が逆になる（c → ac → tac）。

## 問題 loop_04_q03

- 形式: choice
- 知識点: loop_04, pat_02
- 難易度: 2
- キーワード: `for ch in 文字列` `カウント`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
count = 0
for ch in 'hello world':
    if ch == 'o':
        count += 1
print(count)
```

- A. `1`
- B. `2`
- C. `3`
- D. `0`

### 正解

B

### 解説

'hello world' には o が2つある（hell**o** w**o**rld）。

## 問題 loop_04_q04

- 形式: choice
- 難易度: 2
- キーワード: `for ch in 文字列` `空白`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for ch in 'a b':
    print(ch)
```

- A. `a` / `b`
- B. `a` / （空白） / `b`
- C. `a b`
- D. エラーになる

### 正解

B

### 解説

空白も1文字として取り出される。2行目には空白1文字が出力される。

## 問題 loop_04_c01

- 形式: code
- 知識点: loop_04, io_01
- 難易度: 2
- キーワード: `for ch in 文字列` `print`
- タグ: AI作成

### 問題文

英単語を1つ入力として受け取り、その各文字を `-` でつないで1行で出力するプログラムを書け。例：`cat` → `c-a-t-`（最後にも `-` を付けてよい）。

### ヒント

1. `print(ch, end='-')` とすると改行せずに `-` が付く。
2. 最後に `print()` で改行する。

### 解説

`end` を使うと print の後ろに付く文字を変えられる。最後の `-` を付けない書き方は、パターン「区切り付きの文字列の組み立て」（pat_10）で学ぶ。

````python
# @starter
word = input()
# ここにコードを書く
````

````python
# @solution
word = input()
for ch in word:
    print(ch, end='-')
print()
````

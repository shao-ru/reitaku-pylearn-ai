# zip — 練習問題

## 問題 loop_07_q01

- 形式: choice
- 難易度: 2
- キーワード: `zip`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for a, b in zip([1, 2], [3, 4]):
    print(a + b)
```

- A. `3` / `7`
- B. `4` / `6`
- C. `1` / `2` / `3` / `4`
- D. `10`

### 正解

B

### 解説

同じ位置の要素が組になる：1+3=4、2+4=6。

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

for a, b in zip(['p', 'q', 'r'], [1, 2]):

- A. `p 1` / `q 2`

### 正解

### 解説

短い方（2要素）に合わせて終わる。`r` は使われない。

## 問題 loop_07_q03

- 形式: choice

- 難易度: 3
- キーワード: `zip` `if`

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
items = ['ペン', 'ノート', '消しゴム']
stock = [5, 0, 2]
for name, n in zip(items, stock):
    if n == 0:
        print(name)
```

- A. `ペン`
- B. `ノート`
- C. `消しゴム`
- D. 何も出力されない

### 正解

B

### 解説

在庫 0 は 2 番目の要素で、対応する品名は「ノート」。

## 問題 loop_07_q04

- 形式: choice
- 知識点: loop_07, pat_02
- 難易度: 3
- キーワード: `zip` `比較`
- タグ: AI作成

### 問題文

2つのリスト `a` と `b` の同じ位置の要素を比べ、等しい組の数を数えたい。正しいコードを**すべて**選べ。

- A. `sum(1 for x, y in zip(a, b) if x == y)`
- B. `count = 0` のあと `for x, y in zip(a, b):` の中で `if x == y: count += 1`
- C. `for x in a: for y in b: if x == y: count += 1`（count は 0 で初期化）
- D. `count = len(a) == len(b)`

### 正解

A, B

### 解説

C は全組み合わせを比べてしまい、位置が違う要素同士も数える。D は長さが同じかどうかしか調べていない。

## 問題 loop_07_c01

- 形式: code
- 知識点: loop_07, pat_01, io_05
- 難易度: 3
- キーワード: `zip` `合計`
- タグ: AI作成

### 問題文

1行目に商品の単価、2行目に購入個数が、どちらも半角スペース区切りで入力される（同じ個数）。合計金額を出力するプログラムを書け。

### ヒント

1. `list(map(int, input().split()))` で1行を整数のリストにできる。
2. `zip` で単価と個数を組にして、`単価 * 個数` を足していく。

### 解説

zip で2つのリストを組にし、累積（pat_01）で合計する。

````python
# @starter
prices = list(map(int, input().split()))
counts = list(map(int, input().split()))
total = 0
# ここにコードを書く
print(total)
````

````python
# @solution
prices = list(map(int, input().split()))
counts = list(map(int, input().split()))
total = 0
for p, c in zip(prices, counts):
    total += p * c
print(total)
````

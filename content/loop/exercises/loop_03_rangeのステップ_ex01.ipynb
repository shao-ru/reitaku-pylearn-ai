# range のステップ — 練習問題

## 問題 loop_03_q01

- 形式: choice
- 難易度: 2
- キーワード: `range(start, end, step)`
- タグ: AI作成

### 問題文

`range(1, 10, 3)` が生成する数列として正しいものはどれか？

- A. `1, 4, 7`
- B. `1, 4, 7, 10`
- C. `1, 3, 6, 9`
- D. `3, 6, 9`

### 正解

A

### 解説

1 から 3 ずつ増やし、10 の手前で止まる。10 は含まない。

## 問題 loop_03_q02

- 形式: choice
- 難易度: 2
- キーワード: `逆順` `負のステップ`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for i in range(3, 0, -1):
    print(i)
```

- A. `3` / `2` / `1`
- B. `3` / `2` / `1` / `0`
- C. `0` / `1` / `2`
- D. 何も出力されない

### 正解

A

### 解説

3 から 1 ずつ減らし、0 の手前（1）で止まる。

## 問題 loop_03_q03

- 形式: choice
- 難易度: 2
- キーワード: `逆順` `end含まない`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for i in range(5, 0):
    print(i)
```

- A. `5` / `4` / `3` / `2` / `1`
- B. `0` / `1` / `2` / `3` / `4`
- C. 何も出力されない
- D. エラーになる

### 正解

C

### 解説

step を省略すると 1 になる。5 から 1 ずつ増やして 0 の手前まで、という範囲は空なので、ループは1回も実行されない。エラーにはならない。

## 問題 loop_03_q04

- 形式: choice
- 難易度: 3
- キーワード: `range(start, end, step)` `偶数`
- タグ: AI作成

### 問題文

`10` から `2` までの偶数を大きい順に（`10, 8, 6, 4, 2`）生成するものを**すべて**選べ。

- A. `range(10, 1, -2)`
- B. `range(10, 2, -2)`
- C. `range(10, 0, -2)`
- D. `range(2, 11, 2)`

### 正解

A, C

### 解説

負のステップでは end の値そのものは含まれない。`range(10, 2, -2)` は 2 を含まない。D は小さい順になる。

## 問題 loop_03_c01

- 形式: code
- 難易度: 3
- キーワード: `range(start, end, step)` `逆順`
- タグ: AI作成

### 問題文

整数 `n`（`n ≧ 1`）を入力として受け取り、`n` から `1` までのカウントダウンを1行に1つずつ出力し、最後に `Go!` と出力するプログラムを書け。

### ヒント

1. `range` の3つ目の引数に負の数を使う。
2. 1 まで出力したいので、`end` は 0 にする。

### 解説

`range(n, 0, -1)` で n, n-1, …, 1 が得られる。

````python
# @starter
n = int(input())
# ここにコードを書く
````

````python
# @solution
n = int(input())
for i in range(n, 0, -1):
    print(i)
print('Go!')
````

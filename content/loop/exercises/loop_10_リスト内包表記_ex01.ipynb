# リスト内包表記 — 練習問題

## 問題 loop_10_q01

- 形式: choice
- 難易度: 4
- キーワード: `リスト内包表記` `for` `変換`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
result = [x * 2 for x in [1, 2, 3]]
print(result)
```

- A. `[1, 2, 3]`
- B. `[2, 4, 6]`
- C. `[1, 4, 9]`
- D. エラーになる

### 正解

B

### 解説

各要素に2を掛けた新しいリストが生成されるため[2, 4, 6]になる

## 問題 loop_10_q02

- 形式: choice
- 難易度: 4
- キーワード: `リスト内包表記` `if` `フィルタリング`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
nums = [1, 2, 3, 4, 5]
result = [x for x in nums if x > 3]
print(result)
```

- A. `[1, 2, 3]`
- B. `[3, 4, 5]`
- C. `[4, 5]`
- D. `[1, 2]`

### 正解

C

### 解説

x > 3を満たす要素は4と5のみであるため[4, 5]になる

## 問題 loop_10_q03

- 形式: choice
- 難易度: 4
- キーワード: `リスト内包表記` `for` `文字列`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
words = ['apple', 'banana', 'cherry']
result = [w.upper() for w in words]
print(result)
```

- A. `['apple', 'banana', 'cherry']`
- B. `['APPLE', 'BANANA', 'CHERRY']`
- C. `'APPLE BANANA CHERRY'`
- D. エラーになる

### 正解

B

### 解説

upper()で各文字列を大文字に変換した新しいリストが生成される

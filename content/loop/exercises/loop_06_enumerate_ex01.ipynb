# enumerate — 練習問題

## 問題 loop_06_q01

- 形式: choice
- 難易度: 2
- キーワード: `enumerate`
- タグ: AI作成

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for i, c in enumerate(['x', 'y']):
    print(i, c)
```

- A. `0 x` / `1 y`
- B. `1 x` / `2 y`
- C. `x 0` / `y 1`
- D. `(0, 'x')` / `(1, 'y')`

### 正解

A

### 解説

番号は 0 から始まり、`i` に番号、`c` に要素が入る。

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
for i, c in enumerate(['x', 'y'], start=1):

```

- A. `0 x` / `1 y`
- B. `1 x` / `2 y`
- C. `1 y` / `2 x`
- D. エラーになる

### 正解

B

### 解説

`start=1` で番号が 1 から始まる。

## 問題 loop_06_q04

- 知識点: loop_06, cond_01
- 難易度: 3

### 問題文

次のコードを実行したときの出力として正しいものはどれか？

```python
scores = [70, 95, 80]
for i, s in enumerate(scores):
    if s > 90:
        print(i)
```

- A. `95`
- B. `1`
- C. `2`
- D. `0`

### 正解

B

### 解説

95 は 0 から数えて 1 番目なので、`i` は 1。

## 問題 loop_06_c01

- 形式: code
- 知識点: loop_06, pat_05, func_03
- 難易度: 2
- キーワード: `enumerate` `start` `f文字列`
- タグ: AI作成

### 問題文

関数 `numbered(items)` を完成させよ。リスト `items` の各要素の前に 1 から始まる番号を付けた文字列のリストを返す。例：`numbered(['A', 'B'])` → `['1. A', '2. B']`

### ヒント

1. `enumerate(items, start=1)` を使う。
2. `f'{i}. {x}'` のように f 文字列で組み立てる。

### 解説

番号付きのリストを作る典型例。番号を自分で数える変数を用意する必要がない。

````python
# @starter
def numbered(items):
    result = []
    # ここにコードを書く
    return result
````

````python
# @solution
def numbered(items):
    result = []
    for i, x in enumerate(items, start=1):
        result.append(f'{i}. {x}')
    return result
````

# for文の基本構文 — 練習問題

## 問題 loop_01_q01

- 形式: choice
- 難易度: 1
- キーワード: `for` `リスト` `ループ変数`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

```python
fruits = ['apple', 'banana', 'cherry']
for f in fruits:
    print(f)
```

- A. `apple banana cherry`（1行）
- B. `apple` / `banana` / `cherry`（3行）
- C. `f` / `f` / `f`（3行）
- D. エラーになる

### 正解

B

### 解説

リストの各要素が順に変数fに代入されて1行ずつ出力される

## 問題 loop_01_q02

- 形式: choice
- 難易度: 1
- キーワード: `for` `リスト` `ループ変数` `代入`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

```python
items = ['apple', 'banana']
for i in items:
    print(i)
    i = 'grape'
```

- A. `apple` / `banana`
- B. `apple` / `grape`
- C. エラーになる
- D. `grape` / `grape`

### 正解

A

### 解説

ループ変数iを書き換えても次のループでは次のリスト要素が代入されるため元のリストに影響しない

# break と continue — 練習問題

## 問題 loop_09_q01

- 形式: choice
- 難易度: 4
- キーワード: `for` `break` `ループ中断`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
for i in range(5):
    if i == 3:
        break
    print(i)
```

- A. `0` / `1` / `2` / `3`
- B. `0` / `1` / `2`
- C. `3` / `4`
- D. `0` / `1` / `2` / `4`

### 正解

B

### 解説

i==3でbreakするので0,1,2まで出力

## 問題 loop_09_q10

- 形式: choice
- 知識点: loop_09, pat_06
- 難易度: 4
- キーワード: `for` `break` `リスト` `if` `フラグ` `ループ中断`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
scores = [55, 80, 40, 90, 70]
found = False
for s in scores:
    if s < 50:
        found = True
        break
print(found)
```

- A. `False`
- B. `True`
- C. `40`
- D. エラーになる

### 正解

B

### 解説

40<50なのでfound=Trueになってbreak

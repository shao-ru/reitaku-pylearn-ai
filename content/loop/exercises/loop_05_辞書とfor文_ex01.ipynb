# 辞書と for 文 — 練習問題

## 問題 loop_05_q01

- 形式: choice
- 難易度: 3
- キーワード: `for` `dict` `キー`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
d = {'a': 1, 'b': 2, 'c': 3}
for k in d:
    print(k)
```

- A. `1` / `2` / `3`
- B. `a` / `b` / `c`
- C. `a:1` / `b:2` / `c:3`
- D. エラーになる

### 正解

B

### 解説

辞書をfor文で回すとキーが取り出される

## 問題 loop_05_q05

- 知識点: loop_05, cond_01
- 難易度: 3
- キーワード: `for` `dict` `.items()` `if` `条件分岐`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
scores = {'Alice': 90, 'Bob': 55, 'Carol': 75}
for name, score in scores.items():
    if score >= 60:
        print(name)
```

- A. `Alice` / `Bob` / `Carol`
- B. `Alice` / `Carol`
- C. `Bob`
- D. `90` / `75`

### 正解

B

### 解説

score>=60はAlice(90)とCarol(75)

## 問題 loop_05_q07

- 形式: choice
- 知識点: loop_05, pat_05
- 難易度: 3
- キーワード: `for` `dict` `キー` `append` `リスト作成`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
d = {'cat': 3, 'dog': 5, 'bird': 2}
result = []
for k in d:
    result.append(k)
print(result)
```

- A. `[3, 5, 2]`
- B. `['cat', 'dog', 'bird']`
- C. `['cat', 3, 'dog', 5, 'bird', 2]`
- D. エラーになる

### 正解

B

### 解説

キーだけをappendするので['cat','dog','bird']

## 問題 loop_05_q08

- 形式: choice
- 知識点: loop_05, cond_01
- 難易度: 3
- キーワード: `for` `dict` `.items()` `if` `条件分岐`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
stock = {'apple': 5, 'banana': 0, 'cherry': 3}
for item, count in stock.items():
    if count == 0:
        print(item, 'is out of stock')
```

- A. `apple is out of stock`
- B. `banana is out of stock`
- C. `cherry is out of stock`
- D. 何も出力されない

### 正解

B

### 解説

count==0はbananaだけ

## 問題 loop_05_q10

- 形式: choice
- 知識点: loop_05, pat_04
- 難易度: 3
- キーワード: `for` `dict` `.items()` `if` `append` `条件付きリスト作成`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
scores = {'Alice': 90, 'Bob': 55, 'Carol': 75}
high = []
for name, score in scores.items():
    if score >= 70:
        high.append(name)
print(high)
```

- A. `['Alice', 'Bob', 'Carol']`
- B. `['Alice', 'Carol']`
- C. `['Bob']`
- D. `[90, 75]`

### 正解

B

### 解説

score>=70はAlice(90)とCarol(75)

# while 文 — 練習問題

## 問題 loop_11_q01

- 形式: choice
- 難易度: 1
- キーワード: `while` `条件式` `ループ変数`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

```python
count = 1
while count <= 3:
    print(count)
    count += 1
```

- A. 1 / 2 / 3（3行）
- B. 1 / 2（2行）
- C. 1 / 2 / 3 / 4（4行）
- D. count / count / count（3行）

### 正解

A

### 解説

countが1, 2, 3と加算され、3行で順に出力されます。

## 問題 loop_11_q02

- 形式: choice
- 難易度: 1
- キーワード: `while` `無限ループ` `変数の更新`

### 問題文

次のコードを実行したときの説明として正しいものはどれですか？

```python
n = 1
while n < 5:
    print(n)
```

- A. 1 から 5 まで表示されて終了する
- B. 1 から 4 まで表示されて終了する
- C. 1 が無限に出力され続ける（無限ループになる）
- D. エラーが発生する

### 正解

C

### 解説

ループ内でnが更新されないため、条件式 n < 5 が常にTrueとなり無限ループになります。

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

- D. エラーが発生する

## 問題 loop_11_q06

- 難易度: 1
- キーワード: `while` `複合代入演算子` `乗算`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

```python
val = 1
while val < 10:
    print(val)
    val *= 2
```

- A. 1 / 2 / 4 / 8

- C. 2 / 4 / 8
- D. 1 / 3 / 9

A

### 解説

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

## 問題 loop_11_q08

- 形式: choice
- 知識点: loop_11, loop_09
- 難易度: 1
- キーワード: `while` `break` `途中終了`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

```python
n = 1
while n <= 5:
    if n == 3:
        break
    print(n)
    n += 1
```

- A. 1 / 2 / 3
- B. 1 / 2
- C. 1 / 2 / 4 / 5
- D. 3 のみ表示される

### 正解

B

### 解説

nが3のときにbreakで処理が強制終了するため、1と2のみ出力されます。

## 問題 loop_11_q09

- 形式: choice
- 知識点: loop_11, loop_09
- 難易度: 1
- キーワード: `while` `continue` `処理のスキップ`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？

```python
count = 0
while count < 3:
    count += 1
    if count == 2:
        continue
    print(count)
```

- A. 1 / 2 / 3
- B. 1 / 3
- C. 2 のみ表示される
- D. 1 のみ表示される

### 正解

B

### 解説

countが2のときにcontinueによりprint処理がスキップされるため、1と3のみ出力されます。

## 問題 loop_11_q10

- 形式: choice
- 難易度: 1
- キーワード: `while` `リスト` `pop`

### 問題文

次のコードを実行したときの出力結果として正しいものはどれですか？
※ items.pop() はリストの末尾から要素を取り出して削除します。

```python
items = ['A', 'B', 'C']
while items:
    print(items.pop())
```

- A. A / B / C
- B. C / B / A
- C. ['A', 'B', 'C']
- D. 空のリストになるため何も出力されない

### 正解

B

### 解説

pop()は末尾から要素を取り出すため C, B, A の順になり、リストが空（False）になるとループが終了します。

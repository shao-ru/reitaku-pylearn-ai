## 問題 u_loop_for_q01

- 形式: choice
- 知識点: pat_02
- 難易度: 5
- キーワード: `累積カウント` `条件分岐`

### 問題文

テストの点数リストから、60点以上（合格）の人数を数えるコードである。出力として正しいものはどれか？

```python
scores = [45, 72, 88, 55, 91, 63, 49, 78]
pass_count = 0
for s in scores:
    if s >= 60:
        pass_count += 1
print(pass_count)
```

- A. `4`
- B. `5`
- C. `6`
- D. `3`

### 正解

B

### 解説

60点以上の要素は `72, 88, 91, 63, 78` の5つ。ループで1つずつ条件を確認しながら `pass_count` を増やす典型的な**累積カウント**パターン。

## 問題 u_loop_for_q02

- 形式: choice
- 知識点: pat_01, col_05
- 難易度: 5
- キーワード: `タプルのアンパック` `累積合計`

### 問題文

買い物かごの商品名と価格のタプルリストから合計金額を求める。出力として正しいものはどれか？

```python
cart = [("りんご", 150), ("バナナ", 80), ("みかん", 200), ("ぶどう", 350)]
total = 0
for item, price in cart:
    total += price
print(total)
```

- A. `580`
- B. `700`
- C. `780`
- D. `800`

### 正解

C

### 解説

`150 + 80 + 200 + 350 = 780`。`for item, price in cart` のように、タプルの要素を変数に直接アンパックできる。商品名（`item`）は使わず価格（`price`）だけ合算している。

## 問題 u_loop_for_q03

- 形式: choice
- 知識点: pat_03
- 難易度: 5
- キーワード: `最大値` `比較と更新`

### 問題文

`max()` を使わずにリストの最大値を求めるコードである。出力として正しいものはどれか？

```python
values = [3, 17, 8, 42, 15, 6]
max_val = values[0]
for v in values:
    if v > max_val:
        max_val = v
print(max_val)
```

- A. `17`
- B. `6`
- C. `3`
- D. `42`

### 正解

D

### 解説

最初の要素 `3` を暫定最大値として持ち、ループで順に比較・更新していく。`42` が最も大きいため最終的に `max_val = 42` になる。**「先頭を初期値にして更新していく」**パターンはよく使われる。

## 問題 u_loop_for_q04

- 形式: choice
- 知識点: loop_06
- 難易度: 5
- キーワード: `enumerate` `start`

### 問題文

`enumerate()` で順位付けをするコードで、最初に出力される行として正しいものはどれか？

```python
students = ["田中", "鈴木", "佐藤", "山田"]
for i, name in enumerate(students, start=1):
    print(f"{i}位: {name}")
```

- A. `0位: 田中`
- B. `1位: 田中`
- C. `1位: 鈴木`
- D. `田中: 1`

### 正解

B

### 解説

`enumerate(iterable, start=1)` は `(1, 要素)` から始まる連番を生成する。`start` を省略すると 0 始まりになるため注意。最初のループは `i=1, name="田中"` なので `1位: 田中` が出力される。

## 問題 u_loop_for_q05

- 形式: choice
- 知識点: loop_07, cond_01
- 難易度: 5
- キーワード: `zip` `複数リストの同時処理`

### 問題文

`zip()` で名前と点数を組み合わせて、90点以上の人だけ表示するコードである。出力として正しいものはどれか？

```python
names = ["Alice", "Bob", "Carol", "Dave"]
scores = [85, 92, 78, 95]
for name, score in zip(names, scores):
    if score >= 90:
        print(name)
```

- A. `Alice`
- B.

    ```
    Bob
    Dave
    ```
- C. `Carol`
- D. 何も表示されない

### 正解

B

### 解説

`zip(names, scores)` は `("Alice", 85), ("Bob", 92), ("Carol", 78), ("Dave", 95)` のペアを順に生成する。90点以上は Bob（92点）と Dave（95点）の2人なので、2行出力される。

文字列の中に含まれる母音（a, e, i, o, u）の数を数えるコードである。出力として正しいものはどれか？

B

## 問題 u_loop_for_q09

- 形式: choice
- 知識点: loop_09, pat_05
- 難易度: 5
- キーワード: `continue` `偶数判定` `リスト構築`

### 問題文

`continue` を使って偶数のみ二乗した結果をリストに集めるコードである。出力として正しいものはどれか？

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
result = []
for n in numbers:
    if n % 2 != 0:
        continue
    result.append(n * n)
print(result)
```

- A. `[1, 4, 9, 16, 25, 36, 49, 64]`
- B. `[4, 16, 36, 64]`
- C. `[2, 4, 6, 8]`
- D. `[1, 9, 25, 49]`

### 正解

B

### 解説

奇数のとき `continue` でスキップするため、`append` が実行されるのは偶数（2, 4, 6, 8）のみ。それぞれを二乗すると `4, 16, 36, 64`。`continue` は「残りの処理を飛ばして次のループへ」という意味。

## 問題 u_loop_for_q10

- 形式: choice
- 知識点: loop_08, io_01
- 難易度: 5
- キーワード: `二重ループ` `end=" "` `print()`

### 問題文

座席番号を「列・番号」の形式で生成する二重ループのコードで、最初に出力される行として正しいものはどれか？

```python
rows = ["A", "B", "C"]
seats = [1, 2, 3, 4]
for row in rows:
    for seat in seats:
        print(f"{row}{seat}", end=" ")
    print()
```

- A. `A1 A2 A3 A4`
- B. `A1 B1 C1`
- C. `A1 A2 A3`
- D. `1A 2A 3A 4A`

### 正解

A

### 解説

外側ループが `row="A"` のとき、内側ループで `seat` が 1, 2, 3, 4 と回るため `A1 A2 A3 A4` が1行目に出力される。`end=" "` で改行を空白に変え、`print()` で最後に改行している。

## 問題 u_loop_for_q11

- 形式: choice
- 知識点: loop_03, pat_01
- 難易度: 5
- キーワード: `range` `ステップ` `累積合計`

### 問題文

`range` のステップ引数を使って1から100までの奇数の合計を求めるコードである。出力として正しいものはどれか？

```python
total = 0
for i in range(1, 101, 2):
    total += i
print(total)
```

- A. `2500`
- B. `5050`
- C. `2550`
- D. `5000`

### 正解

A

### 解説

`range(1, 101, 2)` は `1, 3, 5, …, 99` の50個の奇数を生成する。奇数の和は `n²`（n=50個）= `2500`。ちなみに `range(1, 101)` で1〜100の全整数の和は `5050`（選択肢B）。

## 問題 u_loop_for_q12

- 形式: choice
- 知識点: pat_04
- 難易度: 5
- キーワード: `フィルタリング` `リスト構築` `append`

### 問題文

体温データから38.0℃以上の値だけを集めるコードである。出力として正しいものはどれか？

```python
temps = [36.2, 37.8, 38.5, 36.9, 39.1, 37.0]
fever = []
for t in temps:
    if t >= 38.0:
        fever.append(t)
print(fever)
```

- A. `[36.2, 36.9, 37.0]`
- B. `[37.8, 38.5, 39.1]`
- C. `[38.5, 39.1]`
- D. `[38.5]`

### 正解

C

### 解説

38.0℃以上の値は `38.5` と `39.1` の2つ。`append` で条件を満たしたものだけを集める**フィルタ+収集**パターン。37.8は38.0未満なので含まれない。

## 問題 u_loop_for_q13

- 形式: choice
- 知識点: loop_09, pat_01
- 難易度: 5
- キーワード: `break` `収束判定` `テイラー展開`

### 問題文

テイラー展開で e（ネイピア数）を近似し、項の値が `1e-6` 未満になったら打ち切るコードである。出力として正しいものはどれか？

```python
total = 0.0
term = 1.0
for n in range(1, 100):
    total += term
    term /= n
    if term < 1e-6:
        print(f"{n}項で打ち切り")
        break
```

- A. `5項で打ち切り`
- B. `8項で打ち切り`
- C. `10項で打ち切り`
- D. `20項で打ち切り`

### 正解

C

### 解説

各項の値は `1, 1, 1/2, 1/6, 1/24, …` と減少していく。`n=10` のとき `term = 1/3628800 ≈ 2.76e-7 < 1e-6` を初めて満たし、`break` でループを抜ける。`break` は「十分な精度に到達したらループを早期終了する」典型的な用途。

## 問題 u_loop_for_q14

- 形式: choice
- 知識点: pat_08
- 難易度: 5
- キーワード: `二重ループ` `組み合わせ` `range`

### 問題文

総当たり対戦表を作るコードで、生成される組み合わせ数として正しいものはどれか？

```python
teams = ["A", "B", "C", "D", "E"]
matches = []
for i in range(len(teams)):
    for j in range(i + 1, len(teams)):
        matches.append((teams[i], teams[j]))
print(len(matches))
```

- A. `10`
- B. `15`
- C. `20`
- D. `25`

### 正解

A

### 解説

5チームから2チームを選ぶ組み合わせ数は `5C2 = 10`。`j = i + 1` から始めることで逆順・同一ペアの重複を防いでいる。内側ループの反復回数は `4 + 3 + 2 + 1 + 0 = 10` 回。

## 問題 u_loop_for_q15

- 形式: choice
- 知識点: pat_06, loop_06
- 難易度: 5
- キーワード: `enumerate` `break` `検索`

### 問題文

センサーの測定値リストから、最初に閾値（100）を超えた値のインデックスを探すコードである。出力として正しいものはどれか？

```python
readings = [42, 67, 85, 91, 103, 78, 115, 88]
found = -1
for i, val in enumerate(readings):
    if val > 100:
        found = i
        break
print(found)
```

- A. `4`
- B. `6`
- C. `5`
- D. `-1`

### 正解

A

### 解説

インデックス4の値が `103` で最初に100を超える。`break` でそこで即座にループを終了するため `found = 4` が出力される。インデックス6の `115` も条件を満たすが、`break` により到達しない。`found = -1` の初期値は「見つからなかった場合」のフォールバック。

# 二重ループ — 練習問題

## 問題 loop_08_q01

- 形式: choice
- 難易度: 4
- キーワード: `for` `二重ループ` `range` `外側` `内側`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
for i in range(2):
    for j in range(2):
        print(i, j)
```

- A. `0 0` / `1 1`
- B. `0 0` / `0 1` / `1 0` / `1 1`
- C. `0 1` / `1 0`
- D. `0 0` / `1 1` / `2 2`

### 正解

B

### 解説

外ループi=0,1、内ループj=0,1の全組み合わせ

- C. `*` / `*` / `*`
- D. `***` / `***` / `***`

A

### 解説

i=1のとき*1個、i=2のとき*2個、i=3のとき*3個

- C. `apple banana` / `1 2`
- D. `apple 1` / `banana 1` / `apple 2` / `banana 2`

B

### 解説

外ループがapple/banana、内ループが1/2

## 問題 loop_08_q09

- 形式: choice
- 知識点: loop_08, pat_01
- 難易度: 4
- キーワード: `for` `二重ループ` `リストのリスト` `累積` `合計`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
matrix = [[1, 2], [3, 4], [5, 6]]
total = 0
for row in matrix:
    for val in row:
        total += val
print(total)
```

- A. `6`
- B. `15`
- C. `21`
- D. `12`

### 正解

C

### 解説

1+2+3+4+5+6=21

## 問題 loop_08_q10

- 形式: choice
- 知識点: loop_08, pat_04
- 難易度: 4
- キーワード: `for` `二重ループ` `if` `append` `タプル` `複合`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
result = []
for i in range(1, 3):
    for j in range(1, 3):
        if i != j:
            result.append((i, j))
print(result)
```

- A. `[(1, 1), (2, 2)]`
- B. `[(1, 2), (2, 1)]`
- C. `[(1, 2)]`
- D. `[(1, 1), (1, 2), (2, 1), (2, 2)]`

### 正解

B

### 解説

i!=jは(1,2)と(2,1)

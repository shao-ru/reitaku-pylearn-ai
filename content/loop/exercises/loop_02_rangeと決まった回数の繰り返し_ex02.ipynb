## 問題 loop_02_c01

- 形式: code
- 難易度: 2
- キーワード: `range(n)` `print`

### 問題文

整数 `n` を1つ入力として受け取り、`0` から `n-1` までの整数を1行に1つずつ出力するプログラムを書け。

### ヒント

1. `range(n)` は `0, 1, ..., n-1` を順に生成する。
2. `for i in range(n):` の中で `print(i)` すればよい。

### 解説

`range(n)` は 0 から始まり n を含まない。`range(1, n)` や `range(n + 1)` と書くと出力される範囲がずれる。

````python
# @starter
n = int(input())
# range を使って 0 から n-1 までを1行ずつ出力する
````

````python
# @solution
n = int(input())
for i in range(n):
    print(i)
````

## 問題 loop_02_c02

- 形式: code
- 難易度: 2
- キーワード: `range(start, end)` `end含まない`

### 問題文

整数 `a` と `b`（`a ≦ b`）がこの順に1行ずつ入力される。`a` から `b` まで（**`b` も含む**）の整数を1行に1つずつ出力するプログラムを書け。

### ヒント

1. `range(start, end)` は `end` を含まない。
2. `b` を含めたいときは `range(a, b + 1)` と書く。

### 解説

`range(a, b)` と書くと `b` が出力されない。「end は含まない」ことを確認する問題。

````python
# @starter
a = int(input())
b = int(input())
# a から b まで（b も含む）を1行ずつ出力する
````

````python
# @solution
a = int(input())
b = int(input())
for i in range(a, b + 1):
    print(i)
````

## 問題 loop_02_q11

- 形式: choice
- 難易度: 2
- キーワード: `range(n)` `range(start, end)`
- タグ: 複数選択

### 問題文

`0` / `1` / `2` を1行ずつ出力するコードを**すべて**選べ。

- A. `for i in range(3): print(i)`
- B. `for i in range(0, 3): print(i)`
- C. `for i in range(1, 3): print(i)`
- D. `for i in [0, 1, 2]: print(i)`

### 正解

A, B, D

### 解説

`range(3)` と `range(0, 3)` はどちらも `0, 1, 2` を生成する。D はリストを直接ループしており、出力は同じ。C は `1` から始まるため `1` / `2` しか出力されない。

## 問題 loop_02_c03

- 形式: code
- 知識点: loop_02, io_03
- 難易度: 2
- キーワード: `range(n)` `input` `int`
- タグ: バグ取り

### 問題文

整数 `n` を入力として受け取り、`0` から `n-1` までを1行ずつ出力するプログラムを書いたが、実行すると次のエラーになる。プログラムを訂正せよ。

```
Traceback (most recent call last):
  File "main.py", line 2, in <module>
    for i in range(n):
TypeError: 'str' object cannot be interpreted as an integer
```

### ヒント

1. エラーメッセージは「`range` に文字列が渡された」と言っている。
2. `input()` が返す値の型は何か。

### 解説

`input()` は常に文字列を返す。`range` には整数が必要なので、`n = int(input())` のように変換する。

````python
# @starter
n = input()
for i in range(n):
    print(i)
````

````python
# @solution
n = int(input())
for i in range(n):
    print(i)
````

## 問題 loop_02_q09

- 形式: choice
- 難易度: 2
- キーワード: `for` `range(start, end)` `n+1` `変数`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
n = 4
for i in range(1, n + 1):
    print(i, 'fish')
```

- A. `1 fish` / `2 fish` / `3 fish`
- B. `1 fish` / `2 fish` / `3 fish` / `4 fish`
- C. `1 fish` / `2 fish` / `3 fish` / `4 fish` / `5 fish`
- D. エラーになる

### 正解

B

### 解説

range(1, 4+1)=range(1,5)なので1,2,3,4 fish

## 問題 loop_02_q10

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
for i in range(2, 8):
    if i % 3 == 0:
        print(i)
```

## 問題 loop_02_q12

- 形式: choice
- 難易度: 2
- キーワード: `for` `range(n)` `回数繰り返し`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
for i in range(4):
    print('!')
```

- A. `!!!!`（1行）
- B. `!` が3行
- C. `!` が4行
- D. `!` が5行

### 正解

C

### 解説

range(4)は0〜3の4回なので'!'が4行

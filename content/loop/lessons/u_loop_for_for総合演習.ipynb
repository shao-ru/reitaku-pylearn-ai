# for文 総合演習

🔑 **キーワード：** `for` `range` `enumerate` `zip` `break` `continue` `items()` `累積` `ネストループ` `リスト生成`

## この知識点について

これまで学んだ `for` 文の要素を組み合わせて、**実際の場面に近い問題**を解く。

### 登場するパターン一覧

| パターン | 用途 | 典型例 |
|---|---|---|
| `for i in range(n)` | 決まった回数のループ | 累積計算、インデックスが必要な処理 |
| `for i in range(a, b, step)` | 範囲・刻み幅の指定 | 偶数のみ、逆順、5刻みなど |
| `for item in list` | リストを順に処理 | 集計、変換、フィルタ |
| `for ch in str` | 文字列を1文字ずつ処理 | 文字カウント、検索、整形 |
| `for k, v in dict.items()` | 辞書のキーと値を同時に取得 | 集計、ランキング、フィルタ |
| `for i, v in enumerate(list, start=1)` | 連番つきで処理 | 順位付け、番号表示 |
| `for a, b in zip(list1, list2)` | 2つのリストを同時に処理 | 名前+点数、座標など |
| `for … if 条件` | 条件つきループ | 合格者のみ、偶数のみ |
| ループ内で `result.append(…)` | リストを逐次構築 | フィルタ結果を集める |
| ネストループ | 2次元データや組み合わせの処理 | 行列、座席、全ペア |
| `break` | 条件を満たしたらループを抜ける | 検索で見つかった瞬間に終了、精度到達 |
| `continue` | 条件を満たす要素だけスキップ | 不要なデータを読み飛ばす |

## 例1：合計金額を求める

````python
cart = [("りんご", 150), ("バナナ", 80), ("みかん", 200)]
total = 0
for item, price in cart:
    total += price
print(total)  # 430
````

タプルのリストをそのままアンパックして受け取れる。

## 例2：enumerate で順位をつける

````python
players = ["Alice", "Bob", "Carol"]
for rank, name in enumerate(players, start=1):
    print(f"{rank}位: {name}")
# 1位: Alice
# 2位: Bob
# 3位: Carol
````

`start=1` を指定すると 1 始まりの連番になる（デフォルトは 0）。

## 例3：条件つきでリストを構築する

````python
temps = [36.2, 37.8, 38.5, 36.9, 39.1, 37.0]
fever = []
for t in temps:
    if t >= 38.0:
        fever.append(t)
print(fever)  # [38.5, 39.1]
````

「条件を満たすものだけ集める」はよく出るパターン。

## 例4：range のステップを使う

````python
# 1から100までの奇数の和
total = 0
for i in range(1, 101, 2):   # 1, 3, 5, ..., 99
    total += i
print(total)  # 2500
````

`range(start, stop, step)` の第3引数でステップ幅を指定できる。

## 例5：break で精度到達後に終了する（テイラー展開）

e（ネイピア数）をテイラー展開で近似する。

$$e = \sum_{n=0}^{\infty} \frac{1}{n!} = 1 + 1 + \frac{1}{2!} + \frac{1}{3!} + \cdots$$

````python
import math

total = 0.0
term = 1.0        # 各項の値（n! の逆数）
threshold = 1e-8  # この値より小さくなったら十分な精度

for n in range(1, 100):
    total += term
    term /= n         # 次の項: 1/n! → 1/(n+1)! = (1/n!) / (n+1)
    if term < threshold:
        print(f"{n}項で収束")
        break

print(f"近似値: {total:.10f}")
print(f"真の値: {math.e:.10f}")

# 12項で収束
# 近似値: 2.7182818262
# 真の値: 2.7182818285
````

**ポイント：** `break` は「条件を満たした瞬間に終了」。無限ループを避けるため最大反復数（ここでは100）を上限として `range` で管理するのが安全。

## 例6：二重ループで全ペアを生成する

````python
# 総当たり対戦表を作る
teams = ["A", "B", "C", "D"]
matches = []
for i in range(len(teams)):
    for j in range(i + 1, len(teams)):
        matches.append((teams[i], teams[j]))

print(matches)
# [('A','B'), ('A','C'), ('A','D'), ('B','C'), ('B','D'), ('C','D')]
````

`j = i + 1` から始めることで、同じ組み合わせの重複を避けている。

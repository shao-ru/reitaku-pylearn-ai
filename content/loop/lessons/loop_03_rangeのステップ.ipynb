# range のステップ

🔑 **キーワード：** `range(start, end, step)` `逆順` `刻み幅`

## 3つ目の引数 step

`range(start, end, step)` は、`start` から `step` ずつ増やした数を、`end` の手前まで生成する。

- `end` を含まないのは `range(start, end)` と同じ
- `step` を省略すると `1` になる

````python
# 0 から 10 未満の偶数
for i in range(0, 10, 2):
    print(i)
````

````python
# 5 刻み
for i in range(5, 26, 5):
    print(i)
````

## 逆順に数える

`step` を負の数にすると、大きい数から小さい数へ数える。このとき `end` は `start` より**小さい**値にする。

````python
# 5, 4, 3, 2, 1
for i in range(5, 0, -1):
    print(i)
````

> ⚠️ `range(5, 0)` のように `step` を省略したまま `start > end` にすると、何も生成されない（ループが1回も実行されない）。

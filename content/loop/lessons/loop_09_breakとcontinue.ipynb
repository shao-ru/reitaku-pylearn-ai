# break と continue

🔑 **キーワード：** `for` `break` `continue` `ループ制御` `スキップ` `中断`

## break — ループを途中で終了する

`break` が実行されると、そのループを**即座に抜け出す**。

```python
for i in range(10):
    if i == 5:
        break
    print(i)
# → 0, 1, 2, 3, 4 まで出力して終了
```

## continue — その回の処理をスキップして次へ

`continue` が実行されると、**その回の残りの処理をスキップ**して次のループに進む。

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
# → 0, 1, 3, 4 が出力される（2はスキップ）
```

## break と continue の違い

| 命令 | 動作 |
|------|------|
| `break` | ループ全体を**終了**する |
| `continue` | その回だけ**スキップ**して次へ進む |

````python
# break の例
for i in range(10):
    if i == 5:
        break
    print(i)
````

````python
# continue の例
for i in range(5):
    if i == 2:
        continue
    print(i)
````

````python
# 実用例：リストの中から目的の値を見つけたらstopする
nums = [3, 7, 1, 9, 4]
for n in nums:
    if n == 9:
        print('見つかった！', n)
        break
    print('探索中:', n)
````

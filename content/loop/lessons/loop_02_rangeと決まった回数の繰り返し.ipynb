# range と決まった回数の繰り返し

🔑 **キーワード：** `for` `range` `range(n)` `range(start, end)` `数列` `回数繰り返し`

## なぜrange関数が必要か？

リストの要素が少なければ `[1, 2, 3]` と直接書けばよいが、1〜100の数列を用意するのは大変。そこで **`range`関数** を使う。

## 2つの使い方

### ① `range(n)` — 0からn未満

```python
range(n)  # 0, 1, 2, ..., n-1
```
例：`range(3)` → `0, 1, 2`

### ② `range(start, end)` — startからend未満

```python
range(start, end)  # start, start+1, ..., end-1
```
例：`range(1, 5)` → `1, 2, 3, 4`

> ⚠️ **endの値は含まれない**ことに注意！nを含めたい場合は `range(1, n+1)` と書く。

````python
# range(n): 0からn-1まで
for i in range(5):
    print(i)
````

````python
# range(start, end): startからend-1まで
for i in range(1, 4):
    print(i, 'sheep')
````

````python
# bを含めたい場合は b+1 にする
a = 2
b = 5
for count in range(a, b + 1):
    print(f'{count} fish')
````

## ① ループ変数を使わないパターン — 単純な回数繰り返し

変数 `i` はリストから値を受け取るが、処理の中で使わないこともある。  
この場合、**「n回だけ同じ処理を繰り返す」** ことが目的。

````python
# 「*」を10回表示する（i は処理の中で使っていない）
for i in range(10):
    print('*')
````

````python
# 「-」を改行なしで10個並べて表示
for i in range(10):
    print('-', end='')
````

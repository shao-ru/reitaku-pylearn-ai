# enumerate

🔑 **キーワード：** `enumerate` `start` `番号付き`

## 番号と要素を同時に受け取る

`enumerate(リスト)` を使うと、ループのたびに **(番号, 要素)** の組が得られる。番号は 0 から始まる。

````python
fruits = ['apple', 'banana', 'cherry']
for i, f in enumerate(fruits):
    print(i, f)
````

## start で始まりの番号を変える

`enumerate(リスト, start=1)` とすると、番号が 1 から始まる。順位や行番号の表示に便利。

````python
players = ['Alice', 'Bob', 'Carol']
for rank, name in enumerate(players, start=1):
    print(f'{rank}位: {name}')
````

> `for i in range(len(リスト)):` と書いて `リスト[i]` で要素を取り出すこともできるが、`enumerate` の方が読みやすく、間違いも少ない。

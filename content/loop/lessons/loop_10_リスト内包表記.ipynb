# リスト内包表記

🔑 **キーワード：** `リスト内包表記` `for` `if` `フィルタリング` `変換`

## 構文

```python
[式 for 変数 in リスト]
[式 for 変数 in リスト if 条件]
```

## 通常の for 文との比較

```python
# 通常の for 文
result = []
for x in [1, 2, 3]:
    result.append(x * 2)

# リスト内包表記（同じ意味）
result = [x * 2 for x in [1, 2, 3]]
```

## ポイント

- for 文より短く書けるが、処理の意味は同じ
- `if` を付けると条件を満たす要素だけを取り出せる（フィルタリング）
- 結果は常に新しいリストになる

````python
# 各要素を2倍にする
numbers = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in numbers]
print(doubled)
````

````python
# 偶数だけを取り出す（フィルタリング）
numbers = [1, 2, 3, 4, 5, 6]
evens = [x for x in numbers if x % 2 == 0]
print(evens)
````

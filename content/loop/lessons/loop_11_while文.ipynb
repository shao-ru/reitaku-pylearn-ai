# while 文

🔑 **キーワード：** `while` `条件式` `無限ループ` `break` `continue`

## 基本構文

```python
while 条件式:
    繰り返し処理1
    繰り返し処理2
    ...
```

## 処理の流れ

1. 処理を実行する前に「条件式」を評価する
2. 条件式が True（真）であれば、whileブロック内の処理を実行する
3. ブロック内の処理が終わったら、再び「条件式」の評価に戻る
4. 条件式が False（偽）になった時点で繰り返しを終了し、whileブロックを抜ける

## ポイント

- 条件式の変化：ループ内で条件式に関わる変数を更新しないと、無限ループになる
- インデント（字下げ）：whileブロック内の処理はインデントで指定する
- for文との違い：繰り返す回数が決まっていない場合（特定の条件を満たすまで続けたい場合）に最適

## コード例1：数値のカウントダウンに対するwhile文

```python
count = 3

while count > 0:
    print('{} sheep'.format(count))
    count -= 1

print('Good night')
```

````python
count = 3

while count > 0:
    print('{} sheep'.format(count))
    count -= 1

print('Good night')
````

## コード例2：条件を満たすまで処理を続ける例

```python
energy = 10

while energy > 0:
    print('Coding in progress... Energy:', energy)
    energy -= 3

print('Out of energy! Time to sleep.')
```

````python
energy = 10

while energy > 0:
    print('Coding in progress... Energy:', energy)
    energy -= 3

print('Out of energy! Time to sleep.')
````

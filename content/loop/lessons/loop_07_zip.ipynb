# zip

🔑 **キーワード：** `zip` `複数のリスト`

## 2つのリストを同時に回す

`zip(リスト1, リスト2)` を使うと、2つのリストから**同じ位置の要素を組にして**取り出せる。

````python
names = ['Alice', 'Bob', 'Carol']
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(name, score)
````

## 長さが違うとき

リストの長さが違う場合は、**短い方**に合わせて終わる。余った要素は使われない。

````python
for a, b in zip([1, 2, 3], ['x', 'y']):
    print(a, b)
````

> 3つ以上のリストも `zip(a, b, c)` のように同時に回せる。

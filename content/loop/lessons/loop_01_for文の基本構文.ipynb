# for文の基本構文

🔑 **キーワード：** `for` `in` `リスト` `ループ変数` `インデント` `forブロック` `繰り返し` `冗長性` `可読性` `保守性`

## なぜ繰り返し文が必要か？

「一部分だけ変えて同じ処理を何度も行う」という場面はプログラムで非常に多い。

例：
- 年ごとにJリーグ選手の平均年俸を計算する
- チームごとに選手の平均年俸を計算する

手作業でコードを書くと：
- コードが長くなる
- ミスが起きやすい
- 変更に弱い（修正箇所が増える）

→ **繰り返し文を使うことでこれらを解決できる！**

````python
# ❌ for文を使わない場合（冗長）
print('1 sheep')
print('2 sheep')
print('3 sheep')
print('4 sheep')
print('Good night')
````

````python
# ✅ for文を使う場合（シンプル・拡張しやすい）
numbers = [1, 2, 3, 4]
for count in numbers:
    print('{} sheep'.format(count))
print('Good night')
````

## 構文

```python
for 変数 in リスト:
    繰り返し処理1
    繰り返し処理2
    ...
```

## 動作の流れ

1. リストの1つ目の値が変数に代入され、forブロックの処理が実行される
2. forブロックが最後まで終わったら、リストの2つ目の値が変数に代入される
3. リストの最後の値まで繰り返す

## ポイント

- ループ変数の名前は何でもよい（わかりやすい名前にする）
- forブロックは **インデント（字下げ）** で指定する
- リストだけでなく、タプル・文字列なども使える

````python
# 数値のリストに対するfor文
numbers = [1, 2, 3, 4]

for count in numbers:
    print('{} sheep'.format(count))

print('Good night')
````

````python
# 文字列のリストに対するfor文
coders = ['yoh', 'miyu', 'ryo', 'airi', 'shun']

for coder in coders:
    print(coder + ' loves coding')
````

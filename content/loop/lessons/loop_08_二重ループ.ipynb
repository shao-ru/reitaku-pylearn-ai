# 二重ループ

🔑 **キーワード：** `for` `二重ループ` `ネスト` `外側のループ` `内側のループ` `インデント`

## 二重ループとは

for文の中にfor文を書くことができる。これを **ネスト（入れ子）** または **二重ループ** という。

```python
for 変数1 in リスト1:      # 外側のループ
    for 変数2 in リスト2:  # 内側のループ
        処理
```

## 動作の流れ

外側のループが1回進むたびに、内側のループが **全部** 実行される。

````python
# 基本例：掛け算
for i in range(1, 4):
    for j in range(1, 4):
        print(i, 'x', j, '=', i * j)
````

````python
# リスト×リストの組み合わせ
colors = ['red', 'blue']
sizes = ['S', 'M', 'L']
for c in colors:
    for s in sizes:
        print(c, s)
````

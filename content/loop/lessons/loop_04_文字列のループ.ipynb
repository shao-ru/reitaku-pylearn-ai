# 文字列のループ

🔑 **キーワード：** `for ch in 文字列` `1文字ずつ`

## 文字列も1文字ずつ取り出せる

`for` 文はリストだけでなく、文字列にも使える。ループ変数には**1文字ずつ**の文字列が入る。

````python
for ch in 'Python':
    print(ch)
````

## 文字を数える

`if` と組み合わせると、特定の文字を探したり数えたりできる。

````python
text = 'banana'
count = 0
for ch in text:
    if ch == 'a':
        count += 1
print(count)
````

> 文字列のインデックス（`text[0]` など）と同じく、取り出される1文字も文字列（`str`）である。

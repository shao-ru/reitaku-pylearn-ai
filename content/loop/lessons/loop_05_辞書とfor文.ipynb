# 辞書と for 文

🔑 **キーワード：** `for` `dict` `辞書` `キー` `値` `.keys()` `.values()` `.items()`

## 辞書（dict）とは

```python
d = {'apple': 100, 'banana': 80, 'cherry': 120}
```

- `'apple'` などが **キー（key）**
- `100` などが **値（value）**
- キーと値のペアで構成される

## for文での3つの使い方

| 書き方 | 取り出すもの |
|--------|-------------|
| `for k in d:` | キーのみ |
| `for k in d.keys():` | キーのみ（明示的） |
| `for v in d.values():` | 値のみ |
| `for k, v in d.items():` | キーと値の両方 |

````python
d = {'apple': 100, 'banana': 80, 'cherry': 120}

# キーだけ取り出す
for k in d:
    print(k)
````

````python
# 値だけ取り出す
for v in d.values():
    print(v)
````

````python
# キーと値の両方を取り出す
for k, v in d.items():
    print(k, ':', v)
````

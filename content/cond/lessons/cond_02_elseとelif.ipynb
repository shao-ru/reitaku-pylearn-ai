# else と elif

🔑 **キーワード：** `else` `elif` `条件の順番`

## else：条件に当てはまらなかったとき

`if` の条件が `False` のときに実行したい処理は `else:` の後に書く。
`if` と `else` のどちらか**一方だけ**が必ず実行される。

````python
age = 17
if age >= 20:
    print('成人です')
else:
    print('未成年です')
````

## elif：条件をいくつも調べる

3つ以上に分けたいときは `elif`（else if の略）を使う。
**上から順に**条件を調べ、最初に `True` になったところだけが実行される。どれにも当てはまらなければ `else` が実行される。

````python
score = 72
if score >= 80:
    print('A')
elif score >= 60:
    print('B')
else:
    print('C')
````

## 条件の順番に注意

`elif` は上から順に調べるので、**条件を書く順番で結果が変わる**。
次のコードは 85 点でも `B` と表示されてしまう。`score >= 60` が先に `True` になるからである。

````python
score = 85
if score >= 60:
    print('B')
elif score >= 80:
    print('A')   # ここには来ない
else:
    print('C')
````

> 範囲で分けるときは、**厳しい条件（大きい値）から先に**書く。

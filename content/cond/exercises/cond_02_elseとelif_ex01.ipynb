## 問題 cond_02-3ceba3

- 形式: choice
- 難易度: 1
- キーワード: `else`
- タグ: AI作成

### 問題文

次のコードの出力はどれか。

```python
x = 7
if x % 2 == 0:
    print('偶数')
else:
    print('奇数')
```

- A. `偶数`
- B. `奇数`
- C. `偶数` と `奇数` の両方
- D. 何も出力されない

### 正解

B

### 解説

`7 % 2` は `1` なので条件 `x % 2 == 0` は `False` になり、`else` の側だけが実行される。

## 問題 cond_02-1ddf32

- 形式: choice
- 難易度: 2
- キーワード: `elif` `条件の順番`
- タグ: AI作成

### 問題文

次のコードの出力はどれか。

```python
x = 15
if x % 15 == 0:
    print('FizzBuzz')
elif x % 3 == 0:
    print('Fizz')
elif x % 5 == 0:
    print('Buzz')
else:
    print(x)
```

- A. `FizzBuzz`
- B. `Fizz`
- C. `FizzBuzz`、`Fizz`、`Buzz` を1行ずつ
- D. `15`

### ヒント

1. `elif` は、最初に `True` になった条件のところだけを実行する。

### 正解

A

### 解説

最初の条件 `x % 15 == 0` が `True` なので `FizzBuzz` を出力し、残りの `elif` と `else` は調べない。15 は 3 でも 5 でも割り切れるが、`Fizz` や `Buzz` は出力されない。

## 問題 cond_02-0b2223

- 形式: choice
- 知識点: cond_02, cond_01
- 難易度: 3
- キーワード: `else` `elif`
- タグ: 複数選択, AI作成

### 問題文

変数 `age` が 20 以上なら `成人`、そうでなければ `未成年` と出力したい。正しく動くコードを**すべて**選べ。

- A.

    ```python
    if age >= 20:
        print('成人')
    else:
        print('未成年')
    ```

- B.

    ```python
    if age >= 20:
        print('成人')
    if age < 20:
        print('未成年')
    ```

- C.

    ```python
    if age > 20:
        print('成人')
    else:
        print('未成年')
    ```

- D.

    ```python
    if age >= 20:
        print('成人')
    elif age < 20:
        print('未成年')
    ```

### ヒント

1. `age` が 19、20、21 のときにそれぞれ何が出力されるか確かめよう。

### 正解

A, B, D

### 解説

C は `age` がちょうど 20 のとき `未成年` と出力してしまう（`>` は「より大きい」で、20 を含まない）。B は `if` を2つ並べているが、2つの条件が同時に `True` になることはないので結果は正しい。ただし、どちらか一方を選ぶ処理は A のように `else` で書くほうが読みやすい。

### ヒント

1. `elif` は上から順に調べる。どの条件を先に書けばよいか考えよう。
2. 「以上」は `>=` で表す。

### 解説

範囲で分けるときは、厳しい条件（`score >= 80`）から先に書く。80 点や 60 点ちょうどのとき（境界の値）も確かめること。

````python
# @starter
score = int(input())
# ここにコードを書く
````

````python
# @solution
score = int(input())
if score >= 80:
    print('A')
elif score >= 60:
    print('B')
else:
    print('C')
````

## 問題 cond_02-ea7380

- 形式: code
- 知識点: cond_02, io_03
- 難易度: 3
- キーワード: `elif` `条件の順番`
- タグ: バグ取り, AI作成

### 問題文

整数を入力し、正なら `正`、0 なら `ゼロ`、負なら `負` と出力したい。
ところが、次のプログラムは `0` を入力すると `ゼロ` ではなく `正` と出力してしまう。正しく動くように直せ。

### ヒント

1. `0` を入力したとき、最初の条件 `n >= 0` は `True` か `False` か。

### 解説

`n >= 0` は 0 のときも `True` になるため、`elif n == 0` まで調べられない。最初の条件を `n > 0` にするか、`n == 0` を先に調べる。

````python
# @starter
n = int(input())
if n >= 0:
    print('正')
elif n == 0:
    print('ゼロ')
else:
    print('負')
````

````python
# @solution
n = int(input())
if n > 0:
    print('正')
elif n == 0:
    print('ゼロ')
else:
    print('負')
````

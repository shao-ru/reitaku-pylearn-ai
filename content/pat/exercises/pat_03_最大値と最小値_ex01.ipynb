# 最大値・最小値の更新 — 練習問題

## 問題 pat_03_q01

- 形式: choice
- 難易度: 3
- キーワード: `for` `リスト` `if` `最大値` `更新`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
nums = [3, 7, 2, 9, 4]
max_val = nums[0]
for n in nums:
    if n > max_val:
        max_val = n
print(max_val)
```

- A. `3`
- B. `7`
- C. `9`
- D. `4`

### 正解

C

### 解説

3,7,2,9,4の中で最大は9

## 問題 pat_03_q05

- 形式: choice
- 難易度: 3
- キーワード: `for` `リスト` `if` `最大値` `最小値` `複合`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
data = [10, 3, 7, 1, 8]
min_val = data[0]
max_val = data[0]
for d in data:
    if d < min_val:
        min_val = d
    if d > max_val:
        max_val = d
print(min_val, max_val)
```

- A. `1 10`
- B. `10 1`
- C. `3 8`
- D. `1 8`

### 正解

A

### 解説

min=1,max=10

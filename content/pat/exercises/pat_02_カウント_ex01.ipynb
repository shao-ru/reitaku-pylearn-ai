# カウント（条件を満たす個数） — 練習問題

## 問題 pat_02_q01

- 形式: choice
- 知識点: pat_02, loop_02
- 難易度: 2
- キーワード: `for` `range` `if` `カウント` `+=` `偶数` `%`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
count = 0
for i in range(10):
    if i % 2 == 0:
        count += 1
print(count)
```

- A. `4`
- B. `5`
- C. `6`
- D. `10`

### 正解

B

### 解説

range(10)で0〜9のうち偶数は0,2,4,6,8の5個

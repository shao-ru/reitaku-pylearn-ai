# 累積（合計・平均） — 練習問題

## 問題 pat_01_c01

- 形式: code
- 知識点: pat_01, loop_02, func_03
- 難易度: 3
- キーワード: `range(start, end)` `累積` `関数`

### 問題文

関数 `total(n)` を完成させよ。`1` から `n` までの整数の合計を返す（`n ≧ 0`、`n = 0` のときは `0`）。`sum()` は使わず、`for` 文と `range` を使うこと。

### ヒント

1. 合計を入れる変数を `0` で初期化しておく。
2. `range(1, n + 1)` で 1 から n までを順に取り出し、`+=` で足していく。

### 解説

range（loop_03）と累積計算（loop_04）を組み合わせる問題。この問題の正誤は両方の知識点の習熟度に反映される。

````python
# @starter
def total(n):
    result = 0
    # ここにコードを書く
    return result
````

````python
# @solution
def total(n):
    result = 0
    for i in range(1, n + 1):
        result += i
    return result
````

## 問題 pat_01_c02

- 形式: code
- 知識点: pat_01, loop_02
- 難易度: 3
- キーワード: `range(start, end)` `累積`
- タグ: 穴埋め

### 問題文

整数 `n` を入力として受け取り、`1` から `n` までの各整数の2乗の和（`1² + 2² + … + n²`）を出力したい。初期コードの `pass` の行を書き換えてプログラムを完成させよ。

### ヒント

1. `total` には、それまでの2乗の和が入っている。
2. `i` の2乗は `i * i` または `i ** 2` で書ける。

### 解説

累積計算の基本形 `total += 値` を、`range(1, n + 1)` のループの中で使う。

````python
# @starter
n = int(input())
total = 0
for i in range(1, n + 1):
    pass  # ここを書き換える
print(total)
````

````python
# @solution
n = int(input())
total = 0
for i in range(1, n + 1):
    total += i * i
print(total)
````

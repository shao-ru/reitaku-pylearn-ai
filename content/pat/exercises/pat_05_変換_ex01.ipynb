# 変換（各要素を加工して新しいリストを作る） — 練習問題

## 問題 col_02_q01

- 形式: choice
- 知識点: col_02
- 難易度: 3
- キーワード: `リスト` `append` `空リスト`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
result = []
result.append(1)
result.append(2)
result.append(3)
print(result)
```

- A. `1 2 3`
- B. `[3]`
- C. `[1, 2, 3]`
- D. エラーになる

### 正解

C

### 解説

appendを3回呼ぶと[1,2,3]になる

## 問題 pat_05_q01

- 形式: choice
- 難易度: 3
- キーワード: `for` `リスト` `append` `演算`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
nums = [1, 2, 3]
result = []
for n in nums:
    result.append(n * 3)
print(result)
```

- A. `[1, 2, 3]`
- B. `[3, 6, 9]`
- C. `[3, 3, 3]`
- D. `9`

### 正解

B

### 解説

各要素×3で[3,6,9]

## 問題 pat_05_q02

- 形式: choice
- 知識点: pat_05, loop_02
- 難易度: 3
- キーワード: `for` `range` `append` `リスト作成`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
result = []
for i in range(1, 5):
    result.append(i)
print(result)
```

- A. `[0, 1, 2, 3, 4]`
- B. `[1, 2, 3, 4]`
- C. `[1, 2, 3, 4, 5]`
- D. `4`

### 正解

B

### 解説

range(1,5)は1〜4なので[1,2,3,4]

## 問題 pat_05_q06

- 形式: choice
- 難易度: 3
- キーワード: `for` `range` `append` `forブロック` `リスト作成`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
result = []
for i in range(1, 4):
    result.append(i)
    result.append(i * 10)
print(result)
```

- A. `[1, 2, 3, 10, 20, 30]`
- B. `[1, 10, 2, 20, 3, 30]`
- C. `[10, 20, 30]`
- D. `[1, 2, 3]`

### 正解

B

### 解説

i=1のとき1と10、i=2のとき2と20、i=3のとき3と30の順でappend

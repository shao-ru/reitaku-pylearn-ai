# フィルタ（条件を満たす要素を集める） — 練習問題

## 問題 pat_04_q01

- 形式: choice
- 難易度: 3
- キーワード: `for` `リスト` `if` `append` `条件付きリスト作成`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
nums = [5, 3, 8, 1]
result = []
for n in nums:
    if n > 4:
        result.append(n)
print(result)
```

- A. `[5, 3, 8, 1]`
- B. `[5, 8]`
- C. `[3, 1]`
- D. `[5, 3]`

### 正解

B

### 解説

4より大きい5と8だけappendされる

## 問題 pat_04_q02

- 形式: choice
- 難易度: 3
- キーワード: `for` `リスト` `if` `len` `append` `条件付きリスト作成`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
names = ['alice', 'bob', 'carol']
result = []
for name in names:
    if len(name) > 3:
        result.append(name)
print(result)
```

- A. `['alice', 'bob', 'carol']`
- B. `['alice', 'carol']`
- C. `['bob']`
- D. `[]`

### 正解

B

### 解説

len>3はalice(5)とcarol(5)。bobは3なので含まない

## 問題 pat_04_q03

- 形式: choice
- 知識点: pat_04, cond_02
- 難易度: 3
- キーワード: `for` `リスト` `if` `else` `append` `偶数` `奇数`

### 問題文

次のコードを実行したときの出力として正しいものはどれですか？

```python
source = [1, 2, 3, 4, 5]
evens = []
odds = []
for n in source:
    if n % 2 == 0:
        evens.append(n)
    else:
        odds.append(n)
print(evens)
print(odds)
```

- A. `[2, 4]` / `[1, 3, 5]`
- B. `[1, 3, 5]` / `[2, 4]`
- C. `[2, 4, 6]` / `[1, 3, 5]`
- D. `[1, 2, 3, 4, 5]` / `[]`

### 正解

A

### 解説

偶数→evens、奇数→odds に振り分けられる

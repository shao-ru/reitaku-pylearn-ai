# 累積（合計・平均）

🔑 **キーワード：** `累積` `+=` `初期化` `合計` `平均` `len`

## ② 累積計算パターン — 合計・積などを求める

**コツ：**
1. ループの**前**に累積用変数を `0` で初期化
2. ループの**中**で `total += 値` と加算
3. ループの**後**で結果を出力

````python
# 1から10までの合計を求める
total = 0
for num in range(1, 11):
    total += num
print(total)
````

````python
# 1² + 2² + ... + 10²
total = 0
for num in range(1, 11):
    total += num ** 2
print(total)
````

## 平均の求め方

合計 ÷ 個数 で求める。合計は累積で、個数は `len()` で取得する。

````python
nums = [70, 85, 60, 90, 75]
total = 0
for n in nums:
    total += n
average = total / len(nums)
print(average)
````

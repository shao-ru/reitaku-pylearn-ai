# 変換（各要素を加工して新しいリストを作る）

🔑 **キーワード：** `for` `リスト` `append` `空リスト` `[]`

## 基本パターン

```python
new_list = []           # ① 空のリストを用意
for i in original:      # ② 元のリストをループ
    new_list.append(処理)  # ③ 結果を追加
print(new_list)         # ④ 完成したリストを確認
```

## `append()` とは

- `リスト.append(値)` → リストの**末尾に**値を追加する
- ループのたびに呼ぶことで、要素を1つずつ積み上げていく

````python
# [1, 2, 3, 4, 5] の各要素を2乗したリストを作る
l = [1, 2, 3, 4, 5]
mylist = []

for i in l:
    mylist.append(i * i)

print(mylist)
````

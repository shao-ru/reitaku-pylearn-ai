# 最大値・最小値の更新

🔑 **キーワード：** `最大値` `最小値` `更新` `初期化`

## 最大値・最小値の求め方

最大値：最初の要素で初期化して、より大きい値が来たら更新する。

````python
nums = [70, 85, 60, 90, 75]

# 最大値
max_val = nums[0]
for n in nums:
    if n > max_val:
        max_val = n
print('最大値:', max_val)

# 最小値
min_val = nums[0]
for n in nums:
    if n < min_val:
        min_val = n
print('最小値:', min_val)
````
